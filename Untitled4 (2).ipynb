```python
import matplotlib.pyplot as plt
import numpy as np

# X aralıkları
x1 = np.linspace(-6, 6, 400)  # Normal sigmoid için geniş aralık
x2 = np.linspace(0, 6, 400)  # Normalleştirilmiş sigmoid için x >= 0 aralığı


# 1. Standart Sigmoid Fonksiyonu
def sigma(x):
  return 1 / (1 + np.exp(-x))


# 2. Gönderdiğin Tanıma Göre Normalleştirilmiş Sigmoid
def normalized_sigmoid_def(x):
  a = sigma(0)  # a = 0.5
  return (sigma(x) - a) / (1 - a)


# Figür oluşturma (1 satır, 2 sütun)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
plt.suptitle(
    "Normal Sigmoid ve Normalleştirilmiş Sigmoid Fonksiyonlarının"
    " Karşılaştırması",
    fontsize=14,
    y=1.02,
)

# --- 1. Grafik: Normal Sigmoid ---
ax1.plot(
    x1,
    sigma(x1),
    color="blue",
    linewidth=2,
    label=r"$\sigma(x) = \frac{1}{1 + e^{-x}}$",
)

ax1.axvline(0, color="gray", linestyle="--")
ax1.set_title("1. Normal Sigmoid Fonksiyonu")
ax1.set_xlabel("x")
ax1.set_ylabel(r"$\sigma(x)$")
ax1.set_ylim(-0.1, 1.1)
ax1.grid(True, linestyle=":", alpha=0.6)
ax1.legend(loc="upper left")

# --- 2. Grafik: Tanımdaki Normalleştirilmiş Sigmoid ---
ax2.plot(
    x2,
    normalized_sigmoid_def(x2),
    color="green",
    linewidth=2,
    label=r"$S(x) = \frac{\sigma(x) - a}{1 - a}$ ",
)


ax2.axhline(0, color="gray", linestyle="--", linewidth=0.8)
ax2.axvline(0, color="gray", linestyle="--", linewidth=0.8)
ax2.set_xlim(-0.5, 6)
ax2.set_ylim(-0.1, 1.1)
ax2.set_title("2. Normalleştirilmiş Sigmoid Fonksiyonu ")
ax2.set_xlabel("x")
ax2.set_ylabel(r"$S(x)$")
ax2.grid(True, linestyle=":", alpha=0.6)
ax2.legend(loc="upper left")

# Genel yerleşim düzeni
plt.tight_layout()
plt.show()
```
