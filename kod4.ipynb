```python
import sympy as sp

# 1. Durum değişkenleri ve parametreler (sembolik)
x, gamma = sp.symbols("x gamma", real=True)
r, K, x_star, eta = sp.symbols("r K x_star eta", positive=True, real=True)
S = sp.Function("S")

# Sistem denklemleri
dx_dt = r * x * (1 - x / K) - gamma * S(x)  # 1. denklem
dgamma_dt = eta * (x - x_star) * S(x)       # 2. denklem

# 2. Adım: 2. denklemi sıfıra eşitleyip x* bulma (eta > 0, S(x) != 0)
x_denge = sp.solve(sp.Eq(dgamma_dt / (eta * S(x)), 0), x)[0]

# 3. Adım: x = x_denge değerini 1. denklemde yerine koyup 0'a eşitleyerek gamma* bulma
dx_dt_yerine_konmus = dx_dt.subs(x, x_denge)
gamma_denge = sp.simplify(sp.solve(sp.Eq(dx_dt_yerine_konmus, 0), gamma)[0])

# 4. Adım: Denge noktası
denge_noktasi = (x_denge, gamma_denge)

print("2. denklemden bulunan x*:", x_denge)
print("1. denklemden bulunan gamma*:", gamma_denge)
print("Denge noktası:", denge_noktasi)

# Doğrulama: Denge noktasında her iki türev de 0 olmalı
print("Denge noktasında dx/dt:", sp.simplify(dx_dt.subs({x: x_denge, gamma: gamma_denge})))
print("Denge noktasında dgamma/dt:", sp.simplify(dgamma_dt.subs({x: x_denge, gamma: gamma_denge})))
```
