```python
import numpy as np


def standart_sigmoid(x, k=0.4):
    """
    Standart lojistik sigmoid fonksiyonu:
    \sigma(x) = 1 / (1 + \exp[-k \cdot x])
    """
    return 1 / (1 + np.exp(-k * x))


def normellestirilmis_sigmoid(x, k=0.4, x0=50):
    """
    Sistemdeki kaydırmalı (shifted) sigmoid tanımı kullanılarak 
    normalleştirilmiş sigmoid S(x) fonksiyonu.
    """
    
    sig = 1 / (1 + np.exp(-k * (x - x0)))
    a = 1 / (1 + np.exp(-k * (0 - x0))) 
    return (sig - a) / (1 - a)

def normellestirilmis_sigmoid_turevi(x, k=0.4, x0=50):
    """
    Normalleştirilmiş sigmoid fonksiyonunun analitik türevi S'(x):
    S'(x) = \sigma'(x) / (1 - a)
    """
    sig = 1 / (1 + np.exp(-k * (x - x0)))
    a = 1 / (1 + np.exp(-k * (0 - x0)))
    sig_prime = k * sig * (1 - sig)
    return sig_prime / (1 - a)


r = 0.8
K = 100
x_star = 50
k = 0.4


S_star = normellestirilmis_sigmoid(x_star, k, x0=x_star)
gamma_star = (r * x_star * (1 - x_star / K)) / S_star


S_prime_star = normellestirilmis_sigmoid_turevi(x_star, k, x0=x_star)
f_prime_star = r * (1 - 2 * x_star / K) - gamma_star * S_prime_star


print("=" * 50)
print(" ÖRNEK DOĞRULAMA VE KARARLILIK ANALİZİ")
print("=" * 50)
print(f"S({x_star})        = {S_star}")
print(f"γ*           = {gamma_star}")
print(f"S'({x_star})       = {S_prime_star}")
print(f"f'({x_star})       = {f_prime_star}")
print("-" * 50)

if f_prime_star < 0:
    print(f"Sonuç: f'({x_star}) = {f_prime_star:.8f} < 0 olduğundan,")
    print(f"x* = {x_star} denge noktası YEREL ASİMPTOTİK KARARLIDIR.")
else:
    print(f"Sonuç: Kararlı değildir.")
print("=" * 50)
```
