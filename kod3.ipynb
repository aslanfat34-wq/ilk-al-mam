```python
import numpy as np


r = 0.8
K = 100
x_star = 50
k = 0.4
eta = 0.05
x0 = 50  


def standart_sigmoid(x):
    return 1 / (1 + np.exp(-k * (x - x0)))

def standart_sigmoid_turevi(x):
    sig = standart_sigmoid(x)
    return k * sig * (1 - sig)


a = standart_sigmoid(0)

def S(x):
    return (standart_sigmoid(x) - a) / (1 - a)

def S_prime(x):
    return standart_sigmoid_turevi(x) / (1 - a)


s_val = S(x_star)
s_prime_val = S_prime(x_star)


gamma_star = (r * x_star * (1 - x_star / K)) / s_val


a_elem = r * (1 - 2 * x_star / K) - gamma_star * s_prime_val


J_star = np.array([
    [a_elem, -s_val],
    [eta * s_val, 0.0]
], dtype=float)


eigenvalues = np.linalg.eigvals(J_star)


print(f"S({x_star}) = {s_val}")
print(f"S'({x_star}) = {s_prime_val}")
print(f"γ*  = {gamma_star}")
print("\nJacobian Matrisi J*:")
print(J_star)
print("\nÖzdeğerler (Eigenvalues):")
print(eigenvalues)
```
