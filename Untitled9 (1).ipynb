```python
import sympy as sp
from IPython.display import display, Math

sp.init_printing(use_latex="mathjax")

# 1. Durum değişkenleri ve parametreler (sembolik)
x, gamma = sp.symbols("x gamma", real=True)
r, K, eta = sp.symbols("r K eta", positive=True, real=True)
x_star = sp.Symbol("x^*", positive=True, real=True)
gamma_star = sp.Symbol(r"\gamma^*", real=True)
S = sp.Function("S")

# Sistem denklemleri:
# 1. denklem: dx/dt = r*x*(1 - x/K) - gamma*S(x)
# 2. denklem: dgamma/dt = eta*(x - x^*)*S(x)
dx_dt = r * x * (1 - x / K) - gamma * S(x)
dgamma_dt = eta * (x - x_star) * S(x)

# 2. Adım: 2. denklemi sıfıra eşitleyerek x* bulma (eta > 0, S(x) > 0)
x_denge = sp.solve(sp.Eq(dgamma_dt / (eta * S(x)), 0), x)[0]

# 3. Adım: x = x* değerini 1. denklemde yerine koyup 0'a eşitleyerek gamma* bulma
dx_dt_yerine_konmus = dx_dt.subs(x, x_denge)
gamma_denge_raw = sp.solve(sp.Eq(dx_dt_yerine_konmus, 0), gamma)[0]
gamma_denge = r * x_star * (1 - x_star / K) / S(x_star)

# Matematiksel çıktı (Jupyter / IPython)
display(Math(r"\text{2. denklemden bulunan } x^*: \quad x = " + sp.latex(x_denge)))
display(Math(r"\text{1. denklemden bulunan } \gamma^*: \quad \gamma^* = " + sp.latex(gamma_denge)))
display(Math(r"\text{Denge noktası: } \quad (x^*, \gamma^*) = \left(" + sp.latex(x_denge) + ",\\," + sp.latex(gamma_denge) + r"\right)"))
```
